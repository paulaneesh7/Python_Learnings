```python
# Text Loader and Directory Loader

from langchain_community.document_loaders import TextLoader, DirectoryLoader

# Text Loader
loader1 = TextLoader("../data/text_files/python_intro.txt")
loader2 = TextLoader("../data/text_files/machine_learning.txt")

doc1 = loader1.load()
doc2 = loader2.load()

print(doc1)
print(doc2)



# Directory Loader
loader3 = DirectoryLoader(
    "../data/text_files",
    glob="**/*.txt",
    loader_cls=TextLoader
)

doc3 = loader3.load()
print(f"Loaded {len(doc3)} documents which are {doc3}")
```
