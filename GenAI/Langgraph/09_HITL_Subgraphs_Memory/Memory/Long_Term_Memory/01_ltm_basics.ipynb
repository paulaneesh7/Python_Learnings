```python
from langgraph.store.memory import InMemoryStore
```

```python
# Create a store
store = InMemoryStore()
```

```python
# creating a namespace
namespace = ("user", "u1")
```

## Creating Memories

```python
# Adding memories
# store.put(namespace, key, value) where key is a unique identifier for the memory and value is the data to be stored.
store.put(namespace, "1", {"data": "User likes pizza"})
store.put(namespace, "2", {"data": "User prefers dark mode"})
```

```python
# Another namespace

namespace2 = ("user", "u2")


# Add memories
store.put(namespace2, "1", {"data": "User likes pasta"})
store.put(namespace2, "2", {"data": "User prefers grid style navigation"})
```

## Retrieving Memories

```python
# store.get(namespace, key) where key is the unique identifier for the memory you want to retrieve.
store.get(namespace, "1")
```

```python
store.get(namespace2, "2")
```

## Retrieving all memories in a namespace

```python
items = store.search(namespace)

for item in items:
    print(item.value)
```

## Semantic searching memories in a namespace

Suppose you want to find memories related to a specific topic or keyword, you can use semantic search. This allows you to retrieve memories that are contextually relevant, even if they don't contain the exact keyword.


Here we have to pass an embedding model and dimensions while creating the memory store. The embedding model is used to convert the text into a vector representation, which can then be used for semantic search. The dimensions parameter specifies the size of the vector representation.

```python
from langchain_openai import OpenAIEmbeddings
from dotenv import load_dotenv

load_dotenv()


embedding_model = OpenAIEmbeddings(
    model="text-embedding-3-small"
)
```

```python
store = InMemoryStore(index={'embed': embedding_model, 'dims': 1536})
```

```python
namespace = ("user", "u1")
```

```python
store.put(namespace, "1", {"data": "User prefers concise answers over long explanations"})
store.put(namespace, "2", {"data": "User likes examples in Python"})
store.put(namespace, "3", {"data": "User usually works late at night"})
store.put(namespace, "4", {"data": "User prefers dark mode in applications"})
store.put(namespace, "5", {"data": "User is learning machine learning"})
store.put(namespace, "6", {"data": "User dislikes overly theoretical explanations"})
store.put(namespace, "7", {"data": "User prefers step-by-step reasoning"})
store.put(namespace, "8", {"data": "User is based in India"})
store.put(namespace, "9", {"data": "User likes real-world analogies"})
store.put(namespace, "10", {"data": "User prefers bullet points over paragraphs"})
```

```python
items = store.search(
    namespace, 
    query="What is the user currently learning?",
    limit=1
)

for item in items:
    print(item.value)
```

```python
items = store.search(
    namespace,
    query="What are user's preferences",
    limit=3
)

for item in items:
    print(item.value)
```
