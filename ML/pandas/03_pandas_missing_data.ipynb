```python
import numpy as np
import pandas as pd
```

### Find Missing Data

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
```

```python
df
```

```python
# Check which values are "Null" value vs which are not
df.isna()
```

```python
# Count the "Null" values
df.isna().sum()
```

```python
# Any missing value in each column
df.isna().any()
```

### Remove Missing Data

```python
df
```

```python
# Removing "Null" values work on rows, so it only leaves the rows where there are no "Null" values
df.dropna()
```

```python
# with thresh=value (this value will define how many not null values are allowed for a row to be considered)
df.dropna(thresh=3)
```

### Filling the Missing Data

```python
# Filling the NaN values only

df.fillna(0)
```

```python
# Fill missing values with different values for each column

values = {'A': 0, 'B': 100, 'C': 300, 'D': 400}

df.fillna(value=values)
```

```python
# Filling the missing values with the mean of that row

df.fillna(df.mean())
```
