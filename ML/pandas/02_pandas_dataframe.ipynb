### Dataframe 
Collection of multiple series is called Dataframe or df.

```python
import numpy as np
import pandas as pd
```

```python
data = {
    'Name': ['John', 'Anna', 'Peter', 'Linda'],
    'Age': [28, 34, 29, 42],
    'City': ['New York', 'Paris', 'Berlin', 'London'],
    'Salary': [65000, 70000, 62000, 85000]
}
```

```python
# Creating a DataFrame

df = pd.DataFrame(data)
```

```python
df
```

```python
data_list = [
    ['John', 28, 'New York', 65000],
    ['Anna', 34, 'Paris', 70000],
    ['Peter', 29, 'Berlin', 62000],
    ['Linda', 42, 'London', 85000]
]
df2 = pd.DataFrame(data_list)
columns = ["Name", "Age", "City", "Salary"]

df2 = pd.DataFrame(data_list, columns=columns)
df2
```

### Remove a column

```python
# If we don't provide inplace=True (then it won't be deleted truly)
# If we don't provide axis = 1, then it won't be searching in the columns for that column_name we provide
# By default it'll search in the rows with axis=0 then it'll give error because er provided it column_name
df.drop("Salary", axis=1, inplace=True)
```

```python
df
```

```python
# Removing 2 cols at once

df.drop(["Age", "City"], axis=1, inplace=True)
```

```python
df
```

```python
# Removing row
df.drop(1, axis=0, inplace=True)
```

```python
df
```

### Conditional Selection

```python
# People whose age is above 30

df2[df2['Age'] > 30]
```

```python
# People whose age is above 30 and City must be Paris

df2[(df2['Age'] > 30) & (df2['City'] == 'Paris')]
```
