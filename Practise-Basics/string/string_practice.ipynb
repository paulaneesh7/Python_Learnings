```python
# 1. Create a string made of the first, middle, and last character

def string_build(str):
    return str[0] + str[int(len(str)/2)] + str[-1]


print(string_build("hello"))
```

```python
# 2. Create a string made of the middle three characters


def string_build_middle(str):
    return str[int(len(str)/2)] + str[int(len(str)/2) + 1] + str[int(len(str)/2) + 2]


print(string_build_middle("hello"))
```

```python
# 3. Append new string in the middle of a given string


def append_string(str1, str2):
    middle_index = int(len(str1)/2)
    new_str = str1[:middle_index] + str2 + str1[middle_index:]

    return new_str


print(append_string("Ault", "Kelly"))
```

```python
# 4. Create a new string made of the first, middle, and last characters of each input string

def string_build(str1, str2):
    first_char1 = str1[0]
    middle_char1 = str1[int(len(str1)/2)]
    last_char1 = str1[-1]
    
    first_char2 = str2[0]
    middle_char2 = str2[int(len(str2)/2)]
    last_char2 = str2[-1]
    
    ans = first_char1 + first_char2 + middle_char1 + middle_char2 + last_char1 + last_char2
    
    return ans

print(string_build("America", "Japan"))
```

```python
# 5. Reverse a string


def reverse_str(s: str) -> str:
    
    s = s[::-1]
    
    return s

print(reverse_str("PYnative"))



# alternative approach
text = "hello"
reversed_text = ''.join(reversed(text))
print(reversed_text)
```

```python
# 6. Find the last position of a given substring


str1 = "Emma is a data scientist who knows Python. Emma works at google."
substring = "Emma"

ans = str1.rfind(substring)

print(ans)
```

```python
# 7. Split a string on hyphens


def split_hypens(str1: str) -> list:
    return str1.split('-')

print(split_hypens("Emma-is-a-data-scientist"))
```

```python
# 8. Find all occurrences of a substring in a given string by ignoring the case


def find_occurrences(str1: str, substring: str):
    
    return str1.lower().count(substring.lower())



print(find_occurrences("Emma is a data scientist who knows Python. Emma works at google.", "Emma"))
print(find_occurrences("Welcome to USA. usa awesome, isn't it?", "usa"))
```

```python
# 9. String characters balance test

def string_balance_test(str1: str, str2: str) -> bool:
    
    flag = True
    for char in str1:
        if char in str2:
            continue
        else:
            flag = False
            break
    return flag


s1 = "yn"
s2 = "PyNative"
print("Is s1 and s2 balanced:", string_balance_test(s1, s2))
```

```python
# 10. Vowel counter


def count_vowels(str1: str) -> int:
    
    vowels = "aeiouAEIOU"
    count = 0
    for char in str1:
        if char in vowels:
            count += 1
    
    return count

print(count_vowels("Emma is a data scientist who knows Python. Emma works at google."))
```

```python
# 11. Prefix/Suffix check

def check_prefix_suffix(str1: str, prefix: str, suffix: str) -> bool:
    ans = str1.startswith(prefix) and str1.endswith(suffix)
    return ans


print(check_prefix_suffix("Emma is a data scientist who knows Python. Emma works at google.", "Emma", "google."))
print(check_prefix_suffix("https://google.com", "https", ".com"))
```

```python
# 12. Swap Case

def swap_case(str1: str) -> str:
    return str1.swapcase()


print(swap_case("Emma is a data scientist who knows Python. Emma works at google."))
```

```python
# 13. Remove whitespace


def remove_whitespace(str1: str) -> str:
    return str1.replace(" ", "")


print(remove_whitespace("Emma is a data scientist who knows Python. Emma works at google."))
```

```python
# 14. Nth character removal from a string

def remove_nth_character(str1: str, n: int) -> str:
    return str1[:n] + str1[n+1:]


print(remove_nth_character("Emma is a data scientist who knows Python. Emma works at google.", 5))
```

```python
# 15. String partitioning


def str_partition(str1: str, separator: str) -> tuple:
    return str1.partition(separator)


print(str_partition("Emma is a data scientist who knows Python. Emma works at google.", "Python"))
```

```python
# 16. File extraction extension

def file_extension_extraction(file_name: str) -> str:
    ans = file_name.split(".")[-1]
    return ans


print(file_extension_extraction("example.txt"))
```

```python
# 17. Lowercase first 


def lowercase_first(str1: str) -> str:
    
    lower = []
    upper = []
    
    for char in str1:
        if char.islower():
            lower.append(char)
        else:
            upper.append(char)
    
    return ''.join(lower + upper)


print(lowercase_first("ArithmeticError"))
```

```python
# 18. Count all letters, digits, and special symbols from a given string

def count_everything(str1: str) -> dict:
    
    letters = 0
    digits = 0
    special_symbols = 0
    
    for char in str1:
        if char.isalpha():
            letters += 1
        elif char.isdigit():
            digits += 1
        else:
            special_symbols += 1
    
    return {"letters": letters, "digits": digits, "special_symbols": special_symbols}



print(count_everything("P@#yn26at^&i5ve"))
```

```python
# 19. Create a mixed string using alternating characters

# Given Input: s1 = "Abc" s2 = "Xyz"

# Expected Output: AzbycX

def mixed_string(str1: str, str2: str) -> str:
    
    n = len(str1)
    mixed = ""
    
    for i in range(n):
        mixed += str1[i] + str2[n - i - 1]
    
    return mixed


print(mixed_string("Abc", "Xyz"))
print(mixed_string("EfG", "MnO"))

```

```python
# 20. Calculate the sum and average of the digits present in a string

def sum_and_average_of_digits(str1: str) -> tuple:
    total_sum = 0
    count = 0
    
    for char in str1:
        if char.isdigit():
            total_sum += int(char)
            count += 1
    
    average: float = total_sum / count if count > 0 else 0
    
    return total_sum, round(average, 2)


print(sum_and_average_of_digits("PYnative29@#8496"))
```

```python
# 21. Count occurrences of all characters within a string

from collections import defaultdict
def count_occurrences(str1: str) -> dict:
    map = defaultdict(int)
    
    for char in str1:
        map[char] += 1
    
    return dict(map)


print(count_occurrences("apple"))
print(count_occurrences("banana"))
```

```python
# 22. Remove empty strings from a list of strings

def remove_empty_strings(str_list: list[str]) -> list[str]:
    ans = [s for s in str_list if s != ""]
    return ans


print(remove_empty_strings(["Emma", "Jon", "", "Kelly", "", "Eric"]))
```

```python
# 23. Remove special symbols/punctuation from a string

def remove_special_symbols(str1: str) -> str:
    ans = ''.join(char for char in str1 if char.isalnum() or char.isspace())
    return ans


print(remove_special_symbols("/*Jon is @developer & musician!!"))
```

```python
# 24. Remove all characters from a string except integers

def remove_non_integers(str1: str) -> str:
    ans = ''.join(char for char in str1 if char.isdigit())
    return ans


print(remove_non_integers("PYnative29@#8496"))
```

```python
# 25. Find words with both letters and numbers

def find_words_with_letters_and_numbers(str1: str) -> None:
    words = str1.split()
    for word in words:
        if any(char.isalpha() for char in word) and any(char.isdigit() for char in word):
            print(word)
            
find_words_with_letters_and_numbers("Emma25 is a data scientist who knows Python3. Emma works at google.")
```
