```python
# 1. Create a string made of the first, middle, and last character

def string_build(str):
    return str[0] + str[int(len(str)/2)] + str[-1]


print(string_build("hello"))
```

```python
# 2. Create a string made of the middle three characters


def string_build_middle(str):
    return str[int(len(str)/2)] + str[int(len(str)/2) + 1] + str[int(len(str)/2) + 2]


print(string_build_middle("hello"))
```

```python
# 3. Append new string in the middle of a given string


def append_string(str1, str2):
    middle_index = int(len(str1)/2)
    new_str = str1[:middle_index] + str2 + str1[middle_index:]

    return new_str


print(append_string("Ault", "Kelly"))
```

```python
# 4. Create a new string made of the first, middle, and last characters of each input string

def string_build(str1, str2):
    first_char1 = str1[0]
    middle_char1 = str1[int(len(str1)/2)]
    last_char1 = str1[-1]
    
    first_char2 = str2[0]
    middle_char2 = str2[int(len(str2)/2)]
    last_char2 = str2[-1]
    
    ans = first_char1 + first_char2 + middle_char1 + middle_char2 + last_char1 + last_char2
    
    return ans

print(string_build("America", "Japan"))
```

```python
# 5. Reverse a string


def reverse_str(s: str) -> str:
    
    s = s[::-1]
    
    return s

print(reverse_str("PYnative"))



# alternative approach
text = "hello"
reversed_text = ''.join(reversed(text))
print(reversed_text)
```

```python
# 6. Find the last position of a given substring


str1 = "Emma is a data scientist who knows Python. Emma works at google."
substring = "Emma"

ans = str1.rfind(substring)

print(ans)
```

```python
# 7. Split a string on hyphens


def split_hypens(str1: str) -> list:
    return str1.split('-')

print(split_hypens("Emma-is-a-data-scientist"))
```

```python
# 8. Find all occurrences of a substring in a given string by ignoring the case


def find_occurrences(str1: str, substring: str):
    
    return str1.lower().count(substring.lower())



print(find_occurrences("Emma is a data scientist who knows Python. Emma works at google.", "Emma"))
print(find_occurrences("Welcome to USA. usa awesome, isn't it?", "usa"))
```

```python
# 9. String characters balance test

def string_balance_test(str1: str, str2: str) -> bool:
    
    flag = True
    for char in str1:
        if char in str2:
            continue
        else:
            flag = False
            break
    return flag


s1 = "yn"
s2 = "PyNative"
print("Is s1 and s2 balanced:", string_balance_test(s1, s2))
```
