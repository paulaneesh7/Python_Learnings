```python
# Basic Operation

# Create set
s = set()


# Add an element
s.add(10)
s.add(20)
s.add(30)


# Remove an element
s.remove(30)
print(s)


# Discard : This method also takes one parameter, which is the item to be removed. 
# If that item is present, it will remove it. It will not throw any error if it is not present.

s.discard(10)
print(s)
```

```python
# 2. Clear all elements in a set


colors = {"red", "green", "blue"}
print("Before clear:", colors)

colors.clear()
print("After clear:", colors)
```

```python
# 3. Find length of a set

s = set()
s.add(10)
s.add(20)
s.add(30)
s.add(40)
s.add(50)

print(len(s))
```

```python
# 4. Check if set is Empty

s = set()
if not s:
    print("The set is empty")
else:
    print("The set is not empty")
```

```python
# 5. Union of Sets


def union_set(set_a, set_b):
    s = set()
    
    for item in set_a:
        s.add(item)
        
    for item in set_b:
        s.add(item)
        
    for i in s:
        print(i, end=' ')
    print()

union_set({1, 2, 3, 4}, {3, 4, 5, 6})


# alternative
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

result = set_a.union(set_b)
print("Union:", result)
```

```python
# 6. Intersection of Set

def set_intersection(set_a, set_b):
    ans = set_a.intersection(set_b)
    
    return ans


print(set_intersection({1, 2, 3, 4}, {3, 4, 5, 6}))
```

```python
# 7. Difference of sets

def diff_set(set_a, set_b):
    ans = set_a.difference(set_b)
    
    return ans


print(diff_set({1, 2, 3, 4}, {3, 4, 5, 6}))
```

```python
# 8. Symmetic Difference

def symmetric_diff_set(set_a, set_b):
    ans = set_a.symmetric_difference(set_b)
    
    return ans


print(symmetric_diff_set({1, 2, 3, 4}, {3, 4, 5, 6}))
```

```python
# 9. Find Max and Min


def find_max(set_a):
    
    max = -10000000000
    min = 100000000000
    
    for item in set_a:
        if item > max:
            max = item
    
    for item in set_a:
        if item < min:
            min = item
    return max, min


max, min = find_max({42, 7, 19, 85, 3, 56})
print(f"Max: {max}")
print(f"Min: {min}")
```

```python
# 10. Sum of set elements

def _sum_elements(set_a):
    sum = 0
    
    for item in set_a:
        sum += item
        
    print(sum)
    
    
_sum_elements({10, 20, 30, 40, 50})
```
