```python
# Basic Operation

# Create set
s = set()


# Add an element
s.add(10)
s.add(20)
s.add(30)


# Remove an element
s.remove(30)
print(s)


# Discard : This method also takes one parameter, which is the item to be removed. 
# If that item is present, it will remove it. It will not throw any error if it is not present.

s.discard(10)
print(s)
```

```python
# 2. Clear all elements in a set


colors = {"red", "green", "blue"}
print("Before clear:", colors)

colors.clear()
print("After clear:", colors)
```
