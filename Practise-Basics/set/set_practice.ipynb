```python
# 5. Union of Sets


def union_set(set_a, set_b):
    s = set()
    
    for item in set_a:
        s.add(item)
        
    for item in set_b:
        s.add(item)
        
    for i in s:
        print(i, end=' ')
    print()

union_set({1, 2, 3, 4}, {3, 4, 5, 6})


# alternative
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

result = set_a.union(set_b)
print("Union:", result)
```

```python
# 6. Intersection of Set

def set_intersection(set_a, set_b):
    ans = set_a.intersection(set_b)
    
    return ans


print(set_intersection({1, 2, 3, 4}, {3, 4, 5, 6}))
```

```python
# 7. Difference of sets

def diff_set(set_a, set_b):
    ans = set_a.difference(set_b)
    
    return ans


print(diff_set({1, 2, 3, 4}, {3, 4, 5, 6}))
```

```python
# 8. Symmetic Difference

def symmetric_diff_set(set_a, set_b):
    ans = set_a.symmetric_difference(set_b)
    
    return ans


print(symmetric_diff_set({1, 2, 3, 4}, {3, 4, 5, 6}))
```

```python
# 9. Find Max and Min


def find_max(set_a):
    
    max = -10000000000
    min = 100000000000
    
    for item in set_a:
        if item > max:
            max = item
    
    for item in set_a:
        if item < min:
            min = item
    return max, min


max, min = find_max({42, 7, 19, 85, 3, 56})
print(f"Max: {max}")
print(f"Min: {min}")
```

```python
# 12. Update with multiple iterable


def update_multiple(s: set, s1: set, tple: tuple, lst: list) -> set:
    s.update(s1)
    s.update(lst)
    s.update(tple)
    
    # simplified approach
    # s.update(s1, tple, lst)
    
    return s


print(update_multiple({1, 2}, {7, 8}, (5, 6), [3, 4]))
```

```python
#  13. Check subset and superset

"""
if Set A is a subset of Set B, then Set B is automatically a superset of Set A.

"""
def check_subset_superset(s1: set, s2: set) -> bool:
    
    flag = s1.issubset(s2) & s2.issuperset(s1)
    
    return flag


print(check_subset_superset({1,2,3}, {1,2,3,4,5}))
```

```python
# 14. Intersection Check with isdisjoint()


def check_isdisjoint(s1: set, s2: set) -> bool:
    ans: bool = s1.isdisjoint(s2)
    
    return ans


print(check_isdisjoint({1,2,3}, {4,5,6}))
```

```python
# 19. The pop operation

def pop_opp(s: set) -> int | str:
    
    try:
        ans = s.pop()
    except KeyError as e:
        return "Error handled!"
    return ans
    

print(pop_opp({100,200,300}))
```

```python
# 20. Filter a set

def filter_set(s: set) -> set:
    ans = {item for item in s if item % 3 == 0}
    
    return ans


print(filter_set({1, 2, 3, 6, 7, 9, 12, 14, 15}))
```

```python
# 21. Find common elements in Lists

def find_common_elements(l1: list, l2: list) -> set:
    s1 = set(l1)
    s2 = set(l2)
    
    ans = s1.intersection(s2)
    
    return ans


print(find_common_elements([1, 2, 3, 4, 5, 3, 2], [3, 4, 5, 6, 7, 4, 5]))
```

```python
# 22. Count Unique words

def count_unique_words(s: str) -> int:
    strng = s.lower()
    st = set()
    for item in strng.split(" "):
        st.add(item)
    
    return len(st)


print(count_unique_words("the cat sat on the mat the cat"))
```

```python
# 23. Convert set to joined string

def set_to_joined_string(s: set) -> str:
    
    strng = ""
    
    for item in s:
        strng += item
        strng += " "
        
    return strng


print(set_to_joined_string({"python", "set", "programming", "tutorial"}))
```

```python
# 24. Proper subset and superset

def proper_subset_superset(a: set, b: set):
    a = {1, 2, 3}
    b = {1, 2, 3, 4, 5}
    print("a is a proper subset of b:", a < b)
    print("b is a proper superset of a:", b > a)
    print("a is a proper subset of a:", a < a)
    
    
proper_subset_superset({1,2,3}, {1,2,3,4,5})
```

```python
# 25. Frozen Set


def show_frozenset(l1: list, l2: list):
    
    fs1 = frozenset(l1)
    fs2 = frozenset(l2)
    
    union_fs = fs1.union(fs2)
    
    intersection_fs = fs1.intersection(fs2)
    
    try:
        fs1.difference_update(fs2)
    except AttributeError as e:
        print(f"Error handled well : {e}")
        fs1 = fs1.difference(fs2)
    
    return intersection_fs, fs1, fs2, union_fs



print(show_frozenset([30,40,50], [60,70,80]))
```

```python
# 27. Remove duplicates 

def remove_duplicates(lst: list) -> list:
    
    s = set()
    for num in lst:
        s.add(num)

    ans = [item for item in s]
    return ans


print(remove_duplicates([3, 1, 4, 1, 5, 9, 2, 6, 5, 3, 5]))
```

```python
# 28. Multi-set difference


def multiset_diff(s1: set, s2: set, s3: set) :
    
    ans = s1.difference(s2, s3)
    
    return ans


print(multiset_diff({1, 2, 3, 4, 5, 6, 7, 8}, {2,4,6}, {5,7,9}))
```

```python
# 30. Shallow Copy vs Assignment

original = {1,23,4,5}

ref = original
ref.add(10)
print("Original", original)
print("Assignment copied", ref)


original = {10,20,30,40,50}
shallow_copy = original.copy()
shallow_copy.add(140)
shallow_copy.add(400)

print("Original", original)
print("Shallow copy", shallow_copy)
```
