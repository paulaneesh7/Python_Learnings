```python
# 1. Write a Python program to add a new key-value pair to a dictionary, 
# modify an existing value, and access a specific key.


# add new key-value to dict
def add_item(dct, key, val):
    dct[key] = val

    return dct


student = {"name": "Alice", "age": 20, "grade": "B"}
print(add_item(student, "city", "kolkata"))


# access key
print(student['age'])

# modify existing val
student['name'] = 'Jedx'
print(student['name'])
```

```python
# 2. Write a Python program to remove a specific key from a dictionary, 
# retrieve all key-value pairs, and check whether a given key exists.


car = {"brand": "Toyota", "model": "Camry", "year": 2022, "color": "blue"}

# remove specific key
car.pop('brand')
print(car)


# retrieve all key-val pairs
for k, v in car.items():
    print(f"Key: {k} and Value: {v}")


# check whether key exists
def check_key_exist(car, k):
    if car.get(k):
        return True
    return False

print(check_key_exist(car, 'age'))
```

```python
# 3. Write a Python program to create a dictionary by mapping two 
# equal-length lists, one containing keys and the other containing 
# values.

def create_dict_from_list(lst1, lst2):
    dict = {}
    for i in range(0, len(lst1)):
        dict[lst1[i]] = lst2[i]

    print(f"Dictionary: {dict}")


print(create_dict_from_list(["name", "age", "city"], ["Bob", 25, "London"]))
```

```python
# 6. Access Nested Dictionary

def access_nested_dict(data):
    new_dict = {}
    for k, v in data.items():
        if isinstance(v, dict):
            new_dict = v
    for k, v in new_dict.items():
        print(f"Key: {k} and Value: {v}")

access_nested_dict({"name": "Alice", "age": 25, "address": {"city": "New York", "state": "NY"}})
```

```python
# 7. Access ‘history’ Key From a Nested Dictionary

def access_history(data):
    
    for k, v in data.items():
        if isinstance(v, dict):
            new_dict = v
            
    for k, v in new_dict.items():
        if k == 'history':
            print(f"Key: {k} and Value: {v}")
            
access_history({"name": "Alice", "age": 25, "address": {"city": "New York", "state": "NY", "history": ["NYC", "LA"]}})
```

```python
# 9. Rename a Key of Dictionary

def rename_dict_key(dct, old_key, new_key):
    if old_key in dct:
        dct[new_key] = dct.pop(old_key)
    
    return dct

print(rename_dict_key({"name": "Alice", "age": 25}, "age", "years"))
```

```python
# 12. Sum total of all the numerical values stored in the dict

def sum_dict(dct):
    total = 0
    for k, v in dct.items():
        if isinstance(v, int):
            total += v
    
    return total


print(sum_dict({"rent": 1200, "food": 300, "transport": 150, "utilities": 200}))
```

```python
# 13. Extract Subset of keys

def create_subset_dict(dct, subset_keys):
    d = {}
    for k, v in dct.items():
        if k in subset_keys:
            d[k] = v
            
    return d
print(create_subset_dict({"id": 42, "username": "jdoe", "email": "jdoe@example.com", "password": "s3cr3t", "joined": "2021-03-15"}, ["id", "username", "email"]))
```

```python
# Merge Two Lists into Dict (zip)


def merge_list_to_dict(lst1, lst2):
    d = {}
    for item1, item2 in zip(lst1, lst2):
        d[item1] = item2
        

    return d


print(merge_list_to_dict(["brand", "model", "year", "color"], ["Honda", "Civic", 2023, "silver"]))
```

```python
# 15. Count Character Frequencies

def count_frequency(text: str)-> dict:
    d = {}
    for char in list(text):
        d[char] = d.get(char, 0)+1
    
    return d
        

count_frequency("hello world")
```

```python
# 16. Update Nested Dictionary

def modify_nested_dict(dct, nested_key, nested_val):
    
    for k, v in dct.items():
        if isinstance(v, dict):
            for ky, val in v.items():
                if ky == nested_key:
                    dct[k][ky] = nested_val
                    
    return dct



print(modify_nested_dict({"name": "TechCorp", "location": {"city": "Berlin", "country": "Germany"}}, "city", "Munich"))
    
```

```python
# 20. Key of Minimum Value


def key_min_value(dct):
    
    # To get the lowest value in a dict
    lowest = min(dct, key=dct.get)
    
    return lowest


print(key_min_value({"apples": 34, "bananas": 12, "oranges": 57, "grapes": 8, "mangoes": 23}))
```

```python
# 21. Key of maximum value

def key_max_value(dct):
    
    # To get the max value from a dict
    maxKey = max(dct, key=dct.get)
    
    return maxKey



print(key_max_value({"Alice": 88, "Bob": 95, "Carol": 72, "Dave": 95, "Eve": 84}))
```

```python
# 22. Python program to convert a list of key-value tuples into a dictionary


def _tuple_to_list(tple)-> dict:
    
    dct = {}
    for item in tple:
        el1 = item[0]
        el2 = item[1]
        
        dct[el1] = el2
        
    return dct

print(_tuple_to_list([("name", "Alice"), ("age", 25), ("city", "Paris")]))
```

```python
# 23. Find common keys

def find_common_keys(dct1, dct2):
    lst = []
    for k, v in dct1.items():
        if k in dct2:
            lst.append(k)
            
    return lst


print(find_common_keys({"a": 1, "b": 2, "c": 3}, {"b": 20, "c": 30, "d": 40}))



# Another method:

d1 = {"a": 1, "b": 2, "c": 3}
d2 = {"b": 20, "c": 30, "d": 40}

common_keys = d1.keys() & d2.keys()

print("Common keys:", common_keys)
```

```python
# 24. Dictionary difference (just use "-" operator over the keys)


def dict_diff(dct1, dct2):
    
    ans = dct1.keys() - dct2.keys()
    
    return ans


print(dict_diff({"a": 1, "b": 2, "c": 3}, {"b": 20, "d": 40}))
```

```python
# 25. Dictionary Intersection (mainly uses "&" operator)


# for only the keys (intersection)
def dict_intersection_keys(dct1, dct2):
    ans = dct1.keys() & dct2.keys()
    
    return ans


# both keys and values intersection
def dict_intersection(dct1, dct2):
    ans = {k:dct1[k] for k in dct1.keys() & dct2.keys() if dct1[k] == dct2[k]}
    
    return ans


print(dict_intersection({"a": 1, "b": 2, "c": 3}, {"a": 1, "b": 99, "c": 3}))
```

```python
# 28. Sort Dictionary by Keys


def sort_dict_keys(dct):
    
    sorted_keys = sorted(dct)
    
    result = {}
    
    for k in sorted_keys:
        result[k] = dct[k]
        
    return result


print(sort_dict_keys({"banana": 3, "apple": 5, "cherry": 1, "date": 4}))
```

```python
# 30. Python program to check whether all values in a dictionary are distinct


def check_dct_unique_values(dct: dict) -> bool:
    
    n = len(list(dct.values()))
    
    check = n == len(set(dct.values()))
    
    return check


print(check_dct_unique_values({"a": 1, "b": 2, "c": 3, "d": 2}))
    
```

```python
# 31 . Python program to verify whether one dictionary is a subset of another or not.

def check_subset_dict(dct1: dict, dct2: dict) -> bool:
    
    for k, v in dct2.items():
        found = False
        for k1, v1 in dct1.items():
            
            if k == k1 and v == v1:
                found = True
                break
        if not found:
            return False
    
    return True


print(check_subset_dict({"a": 1, "b": 2, "c": 3, "d": 4}, {"a": 1, "c": 3}))


main = {"a": 1, "b": 2, "c": 3, "d": 4}
subset = {"a": 1, "c": 3}
is_subset = subset.items() <= main.items()
print(is_subset)

"""

1. subset.items() : Returns a view of the key-value pairs of
the smaller dictionary as a set-like object: {("a", 1), ("c", 3)}.


2. main.items() : Returns a similar view for the larger dictionary. 
Dictionary item views support set operations directly.



3. <= operator : When used on dict item views, it checks whether
every pair in subset.items() also exists in main.item(), returning
True or False




4. Alternative: subset.items().issubset(main.items()) produces 
the same result and may read more clearly in some codebases.

"""
```

```python
# 32. Python program to sort a dictionary's items by the 
# length of their string values, from shortest to longest



def sort_by_str_length(dct: dict) -> dict:
    
    sorted_dict = dict(sorted(dct.items(), key=lambda item: len(item[1])))
    
    return sorted_dict


print(sort_by_str_length({"a": "banana", "b": "kiwi", "c": "strawberry", "d": "fig"}))
        
```

```python
# 33. Key with longest list

def key_longest_list(dct: dict):
    max_len = 0
    
    for k, v in dct.items():
        v_len = len(v)
        if v_len > max_len:
            max_key = k
            max_len = v_len
    
    return max_key


print(key_longest_list({"fruits": ["apple", "banana", "cherry"], "vegs": ["carrot"], "grains": ["rice", "wheat"]}))
```

```python
# 35. Invert to dict

def invert_dict(dct: dict) -> dict:
    
    d = {v: k for k, v in dct.items()}
    
    return d



print(invert_dict({"a": 1, "b": 2, "c": 3}))
```

```python
# 36. Invert with Duplicate Values


def invert_with_duplicate(dct: dict) -> dict:
    
    d = {}
    
    for k, v in dct.items():
        if v not in d:
            d[v] = []
        d[v].append(k)
    
    return d


print(invert_with_duplicate({"a": 1, "b": 2, "c": 1, "d": 3, "e": 2}))
```

```python
# 38. Group by First Letter


def group_by_first_letter(lst: list) -> dict:
    
    lst.sort()
    
    d = {}
    
    for item in lst:
        first_el = item[0]
        d.setdefault(first_el, []).append(item) #setdefault -> key exists : return the value | key missing -> sets default, returns it
        
    print(d)
        
    
    
group_by_first_letter(["apple", "avocado", "banana", "blueberry", "cherry", "apricot"])
```
