```python
# 1. Write a Python program to add a new key-value pair to a dictionary, 
# modify an existing value, and access a specific key.


# add new key-value to dict
def add_item(dct, key, val):
    dct[key] = val

    return dct


student = {"name": "Alice", "age": 20, "grade": "B"}
print(add_item(student, "city", "kolkata"))


# access key
print(student['age'])

# modify existing val
student['name'] = 'Jedx'
print(student['name'])
```

```python
# 2. Write a Python program to remove a specific key from a dictionary, 
# retrieve all key-value pairs, and check whether a given key exists.


car = {"brand": "Toyota", "model": "Camry", "year": 2022, "color": "blue"}

# remove specific key
car.pop('brand')
print(car)


# retrieve all key-val pairs
for k, v in car.items():
    print(f"Key: {k} and Value: {v}")


# check whether key exists
def check_key_exist(car, k):
    if car.get(k):
        return True
    return False

print(check_key_exist(car, 'age'))
```

```python
# 3. Write a Python program to create a dictionary by mapping two 
# equal-length lists, one containing keys and the other containing 
# values.

def create_dict_from_list(lst1, lst2):
    dict = {}
    for i in range(0, len(lst1)):
        dict[lst1[i]] = lst2[i]

    print(f"Dictionary: {dict}")


print(create_dict_from_list(["name", "age", "city"], ["Bob", 25, "London"]))
```

```python
"""
5. Write a Python program to combine two dictionaries into a single dictionary. 
If both dictionaries share a key, the value from the second dictionary 
should take precedence.

"""


def merge_dict(dict1, dict2):
    return dict1 | dict2

print(merge_dict({"a": 1, "b": 2}, {"b": 3, "c": 4}))
```

```python
# 6. Access Nested Dictionary

def access_nested_dict(data):
    new_dict = {}
    for k, v in data.items():
        if isinstance(v, dict):
            new_dict = v
    for k, v in new_dict.items():
        print(f"Key: {k} and Value: {v}")

access_nested_dict({"name": "Alice", "age": 25, "address": {"city": "New York", "state": "NY"}})
```

```python
# 7. Access ‘history’ Key From a Nested Dictionary

def access_history(data):
    
    for k, v in data.items():
        if isinstance(v, dict):
            new_dict = v
            
    for k, v in new_dict.items():
        if k == 'history':
            print(f"Key: {k} and Value: {v}")
            
access_history({"name": "Alice", "age": 25, "address": {"city": "New York", "state": "NY", "history": ["NYC", "LA"]}})
```

```python
# 9. Rename a Key of Dictionary

def rename_dict_key(dct, old_key, new_key):
    if old_key in dct:
        dct[new_key] = dct.pop(old_key)
    
    return dct

print(rename_dict_key({"name": "Alice", "age": 25}, "age", "years"))
```

```python
# 10. Delete a list of keys from a dictionary

product = {"id": 101, "name": "Laptop", "price": 999, "stock": 50, "warehouse": "A3"}
keys_to_remove = ["stock", "warehouse"]

for key in list(product.keys()):
    if key in keys_to_remove:
        del product[key]
        

print(product)

```

```python
# 11. Check Value Existence

def check_value_present(dct, val):
    
    for k, v in dct.items():
        if v in val:
            print(f"'{v}' exists as a value")
            

check_value_present({"alice": "admin", "bob": "editor", "carol": "viewer"}, ['editor', 'manager'])
```

```python
# 12. Sum total of all the numerical values stored in the dict

def sum_dict(dct):
    total = 0
    for k, v in dct.items():
        if isinstance(v, int):
            total += v
    
    return total


print(sum_dict({"rent": 1200, "food": 300, "transport": 150, "utilities": 200}))
```

```python
# 13. Extract Subset of keys

def create_subset_dict(dct, subset_keys):
    d = {}
    for k, v in dct.items():
        if k in subset_keys:
            d[k] = v
            
    return d
print(create_subset_dict({"id": 42, "username": "jdoe", "email": "jdoe@example.com", "password": "s3cr3t", "joined": "2021-03-15"}, ["id", "username", "email"]))
```

```python
# Merge Two Lists into Dict (zip)


def merge_list_to_dict(lst1, lst2):
    d = {}
    for item1, item2 in zip(lst1, lst2):
        d[item1] = item2
        

    return d


print(merge_list_to_dict(["brand", "model", "year", "color"], ["Honda", "Civic", 2023, "silver"]))
```

```python
# 15. Count Character Frequencies

def count_frequency(text: str)-> dict:
    d = {}
    for char in list(text):
        d[char] = d.get(char, 0)+1
    
    return d
        

count_frequency("hello world")
```

```python
# 16. Modify Nested Dictionary

company = {"name": "TechCorp", "location": {"city": "Berlin", "country": "Germany"}}

company["location"]["city"] = "Munich"
print(company)
```

```python
# 16. Update Nested Dictionary

def modify_nested_dict(dct, nested_key, nested_val):
    
    for k, v in dct.items():
        if isinstance(v, dict):
            for ky, val in v.items():
                if ky == nested_key:
                    dct[k][ky] = nested_val
                    
    return dct



print(modify_nested_dict({"name": "TechCorp", "location": {"city": "Berlin", "country": "Germany"}}, "city", "Munich"))
    
```

```python
# 18. Dictionary Comprehension


d = {n: n**2 for n in range(1, 11)}
print(d)
```

```python
def filter_dict(dct, threshold):
    dt = {k: v for k, v in dct.items() if v > threshold}
    
    return dt


print(filter_dict({"Alice": 82, "Bob": 45, "Carol": 91, "Dave": 58, "Eve": 73}, 60))
```

```python
# 20. Key of Minimum Value


def key_min_value(dct):
    
    # To get the lowest value in a dict
    lowest = min(dct, key=dct.get)
    
    return lowest


print(key_min_value({"apples": 34, "bananas": 12, "oranges": 57, "grapes": 8, "mangoes": 23}))
```

```python
# 21. Key of maximum value

def key_max_value(dct):
    
    # To get the max value from a dict
    maxKey = max(dct, key=dct.get)
    
    return maxKey



print(key_max_value({"Alice": 88, "Bob": 95, "Carol": 72, "Dave": 95, "Eve": 84}))
```

```python
# 22. Python program to convert a list of key-value tuples into a dictionary


def _tuple_to_list(tple)-> dict:
    
    dct = {}
    for item in tple:
        el1 = item[0]
        el2 = item[1]
        
        dct[el1] = el2
        
    return dct

print(_tuple_to_list([("name", "Alice"), ("age", 25), ("city", "Paris")]))
```

```python
# 23. Find common keys

def find_common_keys(dct1, dct2):
    lst = []
    for k, v in dct1.items():
        if k in dct2:
            lst.append(k)
            
    return lst


print(find_common_keys({"a": 1, "b": 2, "c": 3}, {"b": 20, "c": 30, "d": 40}))



# Another method:

d1 = {"a": 1, "b": 2, "c": 3}
d2 = {"b": 20, "c": 30, "d": 40}

common_keys = d1.keys() & d2.keys()

print("Common keys:", common_keys)
```
